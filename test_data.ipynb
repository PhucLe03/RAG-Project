```python
contents = """
title
desc

# Section 1

Hello

# Section 2

2's content

# Section 3

3's content

## Subsection

3's sub-content

### subsubsub

3's sub's sub
"""
```

```python
import re

regex = r"^#.*$"
regex = r"(# .*)"
regex = r"(^(?:# .+))"

section_list = re.split(regex, contents, flags=re.MULTILINE)
print(section_list)
```
