````python
# TODO: split document sematically (split at ```, using regex)

import re

def split_text_and_code(document: str):
    # Regex to capture fenced code blocks
    code_block_pattern = r"```(?:[\s\S]*?)```"
    
    # Split the document by the code pattern, capturing code blocks and text separately
    segments = re.split(code_block_pattern, document)
    
    # Iterate through the segments to separate text and code pairs
    text_code_pairs = []
    for segment in segments:
        # Check if it's a code snippet (starts with ``` and ends with ``` after stripping)
        if segment.strip().startswith("```") and segment.strip().endswith("```"):
            # Remove the enclosing backticks from code
            code_snippet = segment.strip("```").strip()
            text_code_pairs.append(('code', code_snippet))
        else:
            # Otherwise, it's normal text
            text_code_pairs.append(('text', segment.strip()))
    
    return text_code_pairs
````
