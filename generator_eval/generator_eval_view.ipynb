```python
from langchain_ollama import OllamaLLM
EVAL_MODEL_NAME = "granite-code:34b"

eval_llm = OllamaLLM(model=EVAL_MODEL_NAME)
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain.prompts import PromptTemplate, ChatPromptTemplate

template = """
You are a strict question-answering evaluator. 
Your responsibility is to score "Answer Relevance" of the `Actual answer` based on `Retrieved context`.
If the `Retrieved context` is not relevant to the `Actual answer`, the score should be 0.
If the `Retrieved context` is irrelevant to the `Actual answer`, the score should be near 0.
If the `Retrieved context` is relevant to the `Actual answer`, the score should be near 1.
If 100 percent of the `Retrieved context` is relevant to the `Actual answer`, the score should be 1.
Just print the score that you calculated in between 0 and 1.

Retrieved context: {context}

Actual answer: {answer}
"""

prompt = PromptTemplate.from_template(template)
chain = prompt | eval_llm | StrOutputParser()
```

```python
eval_result_list = []

def evaluate(context, answer):
    if (type(answer) != str):
        print("Answer is not a string")
        return
    if (len(answer) > 50000):
        print("Answer too long to evaluate")
        return
    eval_result = chain.invoke({
        'context': context,
        'answer': answer
    })
    print(eval_result)
    eval_result_list.append(eval_result)
```

```python
print(MODEL)
for i in range(len(test)):
    print("Testcase", (i+1), "of", len(test))
    evaluate(test['context'][i], test['answer'][i])
```

```python
with open(os.path.join("eval_result", "raw", f"{MODEL}.txt"), "w") as f:
    for eval_result in eval_result_list:
        f.write("Testcase " + str(eval_result_list.index(eval_result) + 1) + " of " + str(len(eval_result_list)) + "\n")
        f.write(eval_result + "\n")
```
